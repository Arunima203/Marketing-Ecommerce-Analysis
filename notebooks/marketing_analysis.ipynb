```python

# Step 5: Inspect All Datasets

datasets = {
    "Customers": customers,
    "Products": products,
    "Campaigns": campaigns,
    "Events": events,
    "Transactions": transactions
}

for name, df in datasets.items():
    print(f"\n===== {name} Dataset =====")
    print("Shape:", df.shape)
    print("\nMissing Values:")
    print(df.isnull().sum())
    print("\nDuplicate Rows:")
    print(df.duplicated().sum())


# Step 6: Remove Duplicate Rows

for name, df in datasets.items():

    before = len(df)

    df.drop_duplicates(inplace=True)

    after = len(df)

    print(f"\n--- {name} ---")
    print("Rows Before Cleaning:", before)
    print("Rows After Cleaning:", after)
    print("Duplicates Removed:", before - after)

# Step 7: Check Data Types

for name, df in datasets.items():

    print(f"\n===== {name} Dataset =====")

    print(df.dtypes)

# Step 8: Convert Date Columns

customers['signup_date'] = pd.to_datetime(customers['signup_date'])

campaigns['start_date'] = pd.to_datetime(campaigns['start_date'])
campaigns['end_date'] = pd.to_datetime(campaigns['end_date'])

events['timestamp'] = pd.to_datetime(events['timestamp'])

transactions['timestamp'] = pd.to_datetime(transactions['timestamp'])

print("Date columns converted successfully!")

# Step 9: Extract Date Information

transactions['year'] = transactions['timestamp'].dt.year

transactions['month'] = transactions['timestamp'].dt.month

transactions['day'] = transactions['timestamp'].dt.day

print(transactions[['timestamp', 'year', 'month', 'day']].head())

# Step 10: Monthly Transaction Analysis

monthly_sales = transactions.groupby('month')['transaction_id'].count()

print(monthly_sales)

monthly_sales = transactions.groupby(['month', 'year'])['transaction_id'].count()

print(monthly_sales)

# Customer Analysis

print("Total Customers:", customers['customer_id'].nunique())

print("\nCustomer Distribution by Country:")
print(customers['country'].value_counts())

print("\nCustomer Distribution by Gender:")
print(customers['gender'].value_counts())

print("\nCustomer Distribution by Loyalty Tier:")
print(customers['loyalty_tier'].value_counts())

# Product Analysis

print("Total Products:", products.shape[0])

print("\nProduct Dataset Columns:")
print(products.columns.tolist())

print("\nFirst 5 Products:")
print(products.head())

print("\nProduct Data Types:")
print(products.dtypes)

# Campaign Analysis

print("Total Campaigns:", campaigns.shape[0])

print("\nCampaign Dataset Columns:")
print(campaigns.columns.tolist())

print("\nFirst 5 Campaigns:")
print(campaigns.head())

print("\nCampaign Data Types:")
print(campaigns.dtypes)

print("\nMissing Values:")
print(campaigns.isnull().sum())

# Step 10: Events Analysis

print("Total Events:", events.shape[0])

print("\nEvent Dataset Columns:")
print(events.columns.tolist())

print("\nFirst 5 Events:")
print(events.head())

print("\nEvent Data Types:")
print(events.dtypes)

print("\nMissing Values:")
print(events.isnull().sum())

print("\nDuplicate Rows:")
print(events.duplicated().sum())

# Step 11: Transactions Analysis

print("Total Transactions:", transactions.shape[0])

print("\nTransaction Dataset Columns:")
print(transactions.columns.tolist())

print("\nFirst 5 Transactions:")
print(transactions.head())

print("\nTransaction Data Types:")
print(transactions.dtypes)

print("\nMissing Values:")
print(transactions.isnull().sum())

print("\nDuplicate Rows:")
print(transactions.duplicated().sum())

print("\nStatistical Summary:")
print(transactions.describe())

print(transactions.columns.tolist())

# Step 8: Campaign Performance Analysis

print("Campaign Columns:")
print(campaigns.columns.tolist())

print("\nCampaign Data:")
display(campaigns.head(10))

print("\nTransaction Columns:")
print(transactions.columns.tolist())

print("\nTransaction Data:")
display(transactions.head(5))

# Step 9: Check Campaign Connection

print("Campaign ID Columns:")

print(campaigns.filter(
    like='campaign'
).columns.tolist())

print("\nTransaction Campaign Columns:")

print(transactions.filter(
    like='campaign'
).columns.tolist())

# Step 10: Check Campaign IDs

print("Campaign Dataset:")
print(campaigns.columns.tolist())

print("\nTransaction Dataset:")
print(transactions.columns.tolist())

print("\nTotal Campaigns:")
print(campaigns['campaign_id'].nunique())

print("\nCampaign IDs in Transactions:")
print(transactions['campaign_id'].nunique())

print("\nTransactions per Campaign:")
print(transactions['campaign_id'].value_counts().head(10))

# Step 11: Campaign-wise Revenue Analysis

campaign_revenue = transactions.groupby('campaign_id').agg(
    total_revenue=('gross_revenue', 'sum'),
    total_transactions=('transaction_id', 'count')
).reset_index()

print("Campaign-wise Revenue:")
display(campaign_revenue.sort_values(
    by='total_revenue',
    ascending=False
))

# Step 12: Merge Campaign and Revenue Data

campaign_performance = campaigns.merge(
    campaign_revenue,
    on='campaign_id',
    how='left'
)

print("Campaign Performance Report:")
display(campaign_performance.head(10))

print("\nMissing Values:")
print(campaign_performance.isnull().sum())

# Step 13: Best Performing Campaign

best_revenue_campaign = campaign_performance.loc[
    campaign_performance['total_revenue'].idxmax()
]

best_transaction_campaign = campaign_performance.loc[
    campaign_performance['total_transactions'].idxmax()
]

print("Best Campaign by Revenue:")
print(best_revenue_campaign)

print("\nBest Campaign by Transactions:")
print(best_transaction_campaign)

# Step 14: Events Analysis

print("Events Dataset Shape:")
print(events.shape)

print("\nEvents Dataset Columns:")
print(events.columns.tolist())

print("\nFirst 10 Events:")
display(events.head(10))

print("\nEvents Dataset Information:")
events.info()

# Step 15: Event Type Analysis

print(events.columns.tolist())

# Count Different Event Types

event_counts = events['event_type'].value_counts()

print("Event Type Distribution:")
print(event_counts)

# Merge Transactions with Customers

customer_transaction = pd.merge(
    transactions,
    customers,
    on='customer_id',
    how='left'
)

print("Customer + Transaction Dataset")
print("Shape:", customer_transaction.shape)

print("\nFirst 5 Rows:")
display(customer_transaction.head())

print("\nMissing Values:")
print(customer_transaction.isnull().sum())

# Merge Customer-Transaction Data with Campaigns

combined_data = pd.merge(
    customer_transaction,
    campaigns,
    on='campaign_id',
    how='left',
    suffixes=('_transaction', '_campaign')
)

print("Final Combined Dataset")
print("Shape:", combined_data.shape)

print("\nFirst 5 Rows:")
display(combined_data.head())

print("\nColumn Names:")
print(combined_data.columns.tolist())



print("Customer Columns:")
print(customers.columns.tolist())

campaign_revenue = combined_data.groupby('campaign_id')[
    'gross_revenue'
].sum().sort_values(ascending=False)

print("Campaign-wise Revenue:")
print(campaign_revenue.head(10))

# Revenue by Marketing Channel

campaign_revenue_channel = combined_data.merge(
    campaigns[['campaign_id', 'channel']],
    on='campaign_id',
    how='left'
)

# Step 20: Revenue by Marketing Channel

channel_revenue = combined_data.groupby(
    'channel'
)['gross_revenue'].sum().sort_values(ascending=False)

print("Revenue by Marketing Channel:")
print(channel_revenue)

channel_revenue.plot(
    kind='bar',
    figsize=(8, 5)
)

plt.title('Revenue by Marketing Channel')
plt.xlabel('Marketing Channel')
plt.ylabel('Total Revenue')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()


# Step 21: Revenue by Campaign Objective

objective_revenue = combined_data.groupby(
    'objective'
)['gross_revenue'].sum().sort_values(ascending=False)

print("Revenue by Campaign Objective:")
print(objective_revenue)

objective_revenue.plot(
    kind='bar',
    figsize=(8, 5)
)

plt.title('Revenue by Campaign Objective')
plt.xlabel('Campaign Objective')
plt.ylabel('Total Revenue')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()


# Step 22: Top 10 Campaigns by Revenue

campaign_revenue = combined_data.groupby(
    'campaign_id'
)['gross_revenue'].sum().sort_values(ascending=False)

print(campaign_revenue.head(10))

campaign_revenue.head(10).plot(
    kind='bar',
    figsize=(10, 5)
)

plt.title('Top 10 Campaigns by Revenue')
plt.xlabel('Campaign ID')
plt.ylabel('Total Revenue')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()

print("Campaign ID 0 Transactions:")
print(transactions[transactions['campaign_id'] == 0].shape)

print("\nCampaign IDs not found in Campaign Dataset:")

missing_campaigns = set(transactions['campaign_id'].unique()) - set(campaigns['campaign_id'].unique())

print(missing_campaigns)

# Step 23: Revenue by Customer Loyalty Tier

loyalty_revenue = combined_data.groupby(
    'loyalty_tier'
)['gross_revenue'].sum().sort_values(ascending=False)

print("Revenue by Customer Loyalty Tier:")
print(loyalty_revenue)

# Visualization

loyalty_revenue.plot(
    kind='bar',
    figsize=(8, 5)
)

plt.title('Revenue by Customer Loyalty Tier')
plt.xlabel('Loyalty Tier')
plt.ylabel('Total Revenue')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()

# Step 24: Revenue by Country

country_revenue = combined_data.groupby(
    'country'
)['gross_revenue'].sum().sort_values(ascending=False)

print("Revenue by Country:")
print(country_revenue)

country_revenue.plot(
    kind='bar',
    figsize=(10, 5)
)

plt.title('Revenue by Country')
plt.xlabel('Country')
plt.ylabel('Total Revenue')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()

# Step 25: Revenue by Gender

gender_revenue = combined_data.groupby(
    'gender'
)['gross_revenue'].sum().sort_values(ascending=False)

print("Revenue by Gender:")
print(gender_revenue)

gender_revenue.plot(
    kind='bar',
    figsize=(7, 5)
)

plt.title('Revenue by Gender')
plt.xlabel('Gender')
plt.ylabel('Total Revenue')
plt.tight_layout()
plt.show()

# Step 26: Product Category Analysis

print("Product Columns:")
print(products.columns.tolist())

# Step 27: Category-wise Revenue

product_data = transactions.merge(
    products,
    on='product_id',
    how='left'
)

category_revenue = product_data.groupby(
    'category'
)['gross_revenue'].sum().sort_values(ascending=False)

print("Revenue by Product Category:")
print(category_revenue)

category_revenue.plot(
    kind='bar',
    figsize=(8, 5)
)

plt.title('Revenue by Product Category')
plt.xlabel('Product Category')
plt.ylabel('Total Revenue')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()

# Step 29: Brand-wise Revenue

brand_revenue = product_data.groupby(
    'brand'
)['gross_revenue'].sum().sort_values(ascending=False)

print("Revenue by Brand:")
print(brand_revenue)

brand_revenue.plot(
    kind='bar',
    figsize=(10, 5)
)

plt.title('Revenue by Brand')
plt.xlabel('Brand')
plt.ylabel('Total Revenue')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()


# Step 30: Premium Product Analysis

premium_revenue = product_data.groupby(
    'is_premium'
)['gross_revenue'].sum()

print("Premium vs Non-Premium Revenue:")
print(premium_revenue)

premium_revenue.plot(
    kind='bar',
    figsize=(7, 5)
)

plt.title('Premium vs Non-Premium Revenue')
plt.xlabel('Is Premium')
plt.ylabel('Total Revenue')
plt.tight_layout()
plt.show()

# Step 31: Event Type Distribution

event_counts = events['event_type'].value_counts()

print("Event Type Distribution:")
print(event_counts)

event_counts.plot(
    kind='bar',
    figsize=(9, 5)
)

plt.title('Distribution of Event Types')
plt.xlabel('Event Type')
plt.ylabel('Number of Events')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()

# Step 32: Most Active Customers

active_customers = events.groupby(
    'customer_id'
).size().sort_values(ascending=False)

print("Top 10 Most Active Customers:")
print(active_customers.head(10))

# Step 33: Customer Event Analysis

customer_events = events.groupby(
    'customer_id'
)['event_type'].nunique().sort_values(ascending=False)

print("Top 10 Customers by Different Event Types:")
print(customer_events.head(10))

# Step 34: Monthly Event Analysis

events['month'] = events['timestamp'].dt.month

monthly_events = events.groupby('month').size()

print("Monthly Events:")
print(monthly_events)

monthly_events.plot(
    kind='line',
    marker='o',
    figsize=(9, 5)
)

plt.title('Monthly Event Trends')
plt.xlabel('Month')
plt.ylabel('Number of Events')
plt.grid(True)
plt.tight_layout()
plt.show()

# Step 35: Refund Analysis

refund_counts = transactions['refund_flag'].value_counts()

print("Refund Distribution:")
print(refund_counts)

refund_revenue = transactions.groupby(
    'refund_flag'
)['gross_revenue'].sum()

print("\nRevenue by Refund Status:")
print(refund_revenue)

# Step 36: Monthly Revenue Trend

monthly_revenue = transactions.groupby(
    ['year', 'month']
)['gross_revenue'].sum()

print("Monthly Revenue:")
print(monthly_revenue)

monthly_revenue.plot(
    kind='line',
    marker='o',
    figsize=(12, 5)
)

plt.title('Monthly Revenue Trend')
plt.xlabel('Year-Month')
plt.ylabel('Total Revenue')
plt.grid(True)
plt.tight_layout()
plt.show()

# Step 37: Campaign Expected Uplift

uplift_analysis = campaigns.groupby(
    'channel'
)['expected_uplift'].mean().sort_values(ascending=False)

print("Average Expected Uplift by Channel:")
print(uplift_analysis)

uplift_analysis.plot(
    kind='bar',
    figsize=(9, 5)
)

plt.title('Expected Campaign Uplift by Channel')
plt.xlabel('Marketing Channel')
plt.ylabel('Average Expected Uplift')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()

# Step 38: Final Business Insights

print("========== BUSINESS INSIGHTS ==========")

print("\n1. Top Revenue Generating Channel:")
print(channel_revenue.idxmax())

print("\n2. Top Revenue Generating Product Category:")
print(category_revenue.idxmax())

print("\n3. Top Revenue Generating Brand:")
print(brand_revenue.idxmax())

print("\n4. Highest Revenue Customer Loyalty Tier:")
print(loyalty_revenue.idxmax())

print("\n5. Highest Revenue Country:")
print(country_revenue.idxmax())

print("\n6. Best Performing Campaign:")
print(campaign_revenue.idxmax())

print("\n========== END OF REPORT ==========")
```
